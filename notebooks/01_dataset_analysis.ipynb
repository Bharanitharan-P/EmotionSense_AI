```python
import os
import pandas as pd
import matplotlib.pyplot as plt
print("Libraries imported successfully!")
```

```python
train_dir = "../dataset/facial-expression-dataset-image-folders-fer2013/data/train"
val_dir = "../dataset/facial-expression-dataset-image-folders-fer2013/data/val"
test_dir = "../dataset/facial-expression-dataset-image-folders-fer2013/data/test"

print("Train Path :", train_dir)
print("Validation Path :", val_dir)
print("Test Path :", test_dir)
```

```python
# Get all emotion class names

emotion_classes = sorted(os.listdir(train_dir))

print("Emotion Classes:")
print(emotion_classes)

print("\nTotal Emotion Classes:", len(emotion_classes))
```

```python
# Count training images in each emotion

train_counts = {}

for emotion in emotion_classes:
    emotion_path = os.path.join(train_dir, emotion)
    train_counts[emotion] = len(os.listdir(emotion_path))

print("Training Images per Emotion\n")

for emotion, count in train_counts.items():
    print(f"{emotion:10} : {count}")
```

```python
# Convert dictionary to DataFrame

df = pd.DataFrame(
    list(train_counts.items()),
    columns=["Emotion", "Count"]
)

print(df)
```

```python
# Draw bar chart


plt.figure(figsize=(10,5))

plt.bar(
    df["Emotion"],
    df["Count"],
    edgecolor="black"
)

plt.xticks(rotation=20)

plt.title("Training Images per Emotion")
plt.xlabel("Emotion")
plt.ylabel("Number of Images")

plt.tight_layout()
plt.show()
```

```python
import random
from PIL import Image

# Select one random image from each emotion

plt.figure(figsize=(15,8))

for i, emotion in enumerate(emotion_classes):

    emotion_path = os.path.join(train_dir, emotion)

    image_name = random.choice(os.listdir(emotion_path))

    image_path = os.path.join(emotion_path, image_name)

    image = Image.open(image_path)

    plt.subplot(2,4,i+1)
    plt.imshow(image)
    plt.title(f"Class {emotion}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
