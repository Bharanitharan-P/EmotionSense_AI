```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os

print("Libraries Imported Successfully!")
```

```python
dataset_path = "../dataset/facial-expression-dataset-image-folders-fer2013/data"

train_dir = os.path.join(dataset_path, "train")
val_dir = os.path.join(dataset_path, "val")
test_dir = os.path.join(dataset_path, "test")

print("Train:", train_dir)
print("Validation:", val_dir)
print("Test:", test_dir)
```

```python
print(os.path.exists(train_dir))
print(os.path.exists(val_dir))
print(os.path.exists(test_dir))
```

```python
train_datagen = ImageDataGenerator()

print("Training Data Generator Created!")
```

```python
val_datagen = ImageDataGenerator()

print("Validation Data Generator Created!")
```

```python
test_datagen = ImageDataGenerator()

print("Test Data Generator Created!")
```

```python
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode="categorical",
    shuffle=True
)

print("✅ Training Generator Created!")
```

```python
val_generator = val_datagen.flow_from_directory(
    val_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode="categorical",
    shuffle=False
)

print("✅ Validation Generator Created!")
```

```python
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode="categorical",
    shuffle=False
)

print("✅ Test Generator Created!")
```

```python
images, labels = next(train_generator)

print("One Batch Loaded Successfully!")
```

```python
print("Image Batch Shape :", images.shape)
print("Label Batch Shape :", labels.shape)
```
