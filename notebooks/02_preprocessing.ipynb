```python
import os
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image
```

```python
train_dir = "../dataset/facial-expression-dataset-image-folders-fer2013/data/train"

print(train_dir)
```

```python
emotion_label = "Happy"
emotion = "3"

emotion_path = os.path.join(train_dir, emotion)

image_name = os.listdir(emotion_path)[0]

image_path = os.path.join(emotion_path, image_name)

image = Image.open(image_path)

image
```

```python
print("Image Size :", image.size)
print("Image Mode :", image.mode)
print("Image Format :", image.format)
```

```python
image_array = np.array(image)

print(image_array)
```

```python
print("Shape :", image_array.shape)
print("Data Type :", image_array.dtype)
print("Minimum Pixel :", image_array.min())
print("Maximum Pixel :", image_array.max())
```

```python
# Resize image

resized_image = image.resize((224, 224))

print("Original Size :", image.size)
print("Resized Size :", resized_image.size)

resized_image
```

```python
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.imshow(image, cmap="gray")
plt.title("Original (48x48)")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(resized_image, cmap="gray")
plt.title("Resized (224x224)")
plt.axis("off")

plt.show()
```

```python
# Convert grayscale to RGB

rgb_image = resized_image.convert("RGB")

print("Image Mode :", rgb_image.mode)

rgb_image
```

```python
rgb_array = np.array(rgb_image)

print("Shape :", rgb_array.shape)
print("Data Type :", rgb_array.dtype)
```

```python
# Normalize image

normalized_image = rgb_array / 255.0

print("Minimum Value :", normalized_image.min())
print("Maximum Value :", normalized_image.max())
```

```python
print(normalized_image)
```

```python
print("Shape :", normalized_image.shape)
print("Data Type :", normalized_image.dtype)
```

```python
import os

dataset_path = "../dataset/facial-expression-dataset-image-folders-fer2013/data/train"

print("Training Images Per Class\n")

total = 0

for emotion in sorted(os.listdir(dataset_path)):
    emotion_path = os.path.join(dataset_path, emotion)

    if os.path.isdir(emotion_path):
        count = len(os.listdir(emotion_path))
        total += count
        print(f"{emotion:<10} : {count}")

print("\nTotal Images:", total)
```
